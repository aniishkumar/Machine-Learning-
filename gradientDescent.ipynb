```python
def compute_cost(x,y,w,b):
    size =x.shape[0]
    cost=0

    for i in range(size):
        f_wb = w*x[i]+b
        cost += (f_wb-y[i])**2
    total_cost = 1/(2*size) * cost
    return total_cost    
```
